# Train YOLOv8 Chair Detector
This notebook trains a custom YOLOv8 model using your chair dataset.

### Step 1: Turn on GPU
Go to **Runtime > Change runtime type** in the top menu, and select **T4 GPU**.

```
!pip install ultralytics roboflow
```

### Step 2: Upload Dataset
Compress your local `chair_detection_dataset` folder into a `.zip` file.
Drag and drop the `.zip` file into the file browser on the left menu of this Colab page.

```
!unzip -q chair_detection_dataset.zip -d chair_detection_dataset
```

### Step 3: Train the Model
Run the cell below to start training. It will take a few minutes.

```
from ultralytics import YOLO

# Load YOLOv8 small model
model = YOLO('yolov8s.pt')

# Train the model on your dataset
results = model.train(
    data='/content/chair_detection_dataset/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    name='custom_chair_detector'
)
```

### Step 4: Download the Trained Model
Run this cell to download your trained `best.pt` file to your computer.

```
from google.colab import files

# Download the best weights
files.download('/content/runs/detect/custom_chair_detector/weights/best.pt')
```
